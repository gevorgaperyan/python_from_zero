## 3️⃣ Lists, tuples, sets, dictionaries (collection)

| Type | Syntax | Ordered | Duplicates | **Mutable?** | Use it for |
|---|---|---|---|---|---|
| **list** | `[1, 2, 3]` | ✅ | ✅ | ✅ yes | A changeable sequence (like a column) |
| **tuple** | `(1, 2, 3)` | ✅ | ✅ | ❌ **immutable** | Fixed data, e.g. coordinates |
| **set** | `{1, 2, 3}` | ❌ | ❌ (auto-removed) | ✅ yes | Unique values, `DISTINCT` |
| **dict** | `{"a": 1}` | ✅ | keys unique | ✅ yes | Key → value lookups (like a row/JSON) |

> **Mutable** = can be changed after creation. **Immutable** = cannot.

````python
# LIST – mutable
cities = ["Yerevan", "Gyumri", "Vanadzor"]
cities.append("Dilijan")        # add
cities[0] = "Erevan"            # change
cities.remove("Gyumri")         # delete
print("list :", cities)
````

````python
# TUPLE – immutable
point = (40.18, 44.51)
print("tuple:", point)
try:
    point[0] = 41               # ❌ not allowed
except TypeError as e:
    print("Error ->", e)
````

````python
# SET – unique values only (just like SELECT DISTINCT)
countries = ["AM", "DE", "AM", "FR", "DE", "AM"]
unique_countries = set(countries)
print("set  :", unique_countries)
print("list(set(...)) :", sorted(unique_countries))
````

````python
# DICTIONARY – key/value pairs
employee = {"id": 101, "name": "Anna", "dept": "Sales"}
print(employee["name"])               # read
employee["salary"] = 3250             # add
employee["dept"] = "Marketing"        # change
print(employee)
print(employee.keys(), employee.values())
````

### 🧠 Mutable vs immutable — a classic beginner trap

````python
a = [1, 2, 3]
b = a            # b is NOT a copy, it's the SAME list
b.append(99)
print("a =", a)  # a changed too!

c = a.copy()     # a real copy
c.append(100)
print("a =", a, "| c =", c)
````

## 4️⃣ Slicing

Syntax: `sequence[start : stop : step]` — **start is included, stop is excluded**, counting from **0**.

```
 index:    0    1    2    3    4    5
 value:   'P'  'y'  't'  'h'  'o'  'n'
 negative: -6   -5   -4   -3   -2   -1
```

````python
nums = [10, 20, 30, 40, 50, 60, 70]

print(nums[0])        # first            -> 10
print(nums[-1])       # last             -> 70
print(nums[1:4])      # index 1,2,3      -> [20, 30, 40]
print(nums[:3])       # first three      -> [10, 20, 30]
print(nums[4:])       # from index 4     -> [50, 60, 70]
print(nums[::2])      # every 2nd        -> [10, 30, 50, 70]
print(nums[::-1])     # reversed         -> [70, 60, ... 10]

word = "JupyterLab"
print(word[:7], "|", word[-3:])   # slicing works on strings too
````

## 5️⃣ Excel → Pandas

**Pandas** gives you the `DataFrame` — basically *an Excel sheet or SQL table inside Python*.

First, let's create a realistic sample file `sales.xlsx` (with a few deliberate duplicates!). In real life you'd already have your own file.

````python
import pandas as pd
import numpy as np
from pathlib import Path

np.random.seed(42)
n = 40
raw = pd.DataFrame({
    "OrderID":   range(1001, 1001 + n),
    "Date":      pd.date_range("2026-01-01", periods=n, freq="3D"),
    "Customer":  np.random.choice(["Anna", "Boris", "Carla", "David", "Elena"], n),
    "Region":    np.random.choice(["North", "South", "East", "West"], n),
    "Product":   np.random.choice(["Laptop", "Phone", "Tablet", "Monitor"], n),
    "Quantity":  np.random.randint(1, 6, n),
    "UnitPrice": np.random.choice([1200, 800, 450, 300], n),
})
# add 3 duplicate rows to demonstrate cleaning
raw = pd.concat([raw, raw.iloc[[2, 7, 15]]], ignore_index=True)
raw.to_excel("sales.xlsx", index=False, sheet_name="Orders")
print("Created:", Path("sales.xlsx").resolve())
````

## 8️⃣ SQL `SELECT` → Pandas

To compare directly, we load our data into a small **SQLite database** and run the *same question* both ways.

````python
import sqlite3
clean = df.drop_duplicates().copy()
clean["Total"] = clean["Quantity"] * clean["UnitPrice"]

con = sqlite3.connect(":memory:")                     # temporary database in RAM
clean.to_sql("orders", con, index=False, if_exists="replace")
print("SQL table 'orders' ready")
````

| SQL | Pandas |
|---|---|
| `SELECT * FROM orders` | `df` |
| `SELECT Customer, Total FROM orders` | `df[["Customer", "Total"]]` |
| `SELECT TOP 5 * FROM orders` | `df.head(5)` |
| `SELECT DISTINCT Region FROM orders` | `df["Region"].unique()` |
| `SELECT Total*0.2 AS Tax` | `df["Tax"] = df["Total"] * 0.2` |
| `ORDER BY Total DESC` | `df.sort_values("Total", ascending=False)` |

````python
# ── SQL ──
sql = "SELECT Customer, Product, Total FROM orders ORDER BY Total DESC LIMIT 5"
display(pd.read_sql(sql, con))

# ── Pandas ──
clean[["Customer", "Product", "Total"]].sort_values("Total", ascending=False).head(5)
````

## 9️⃣ SQL `WHERE` → Pandas filtering

| SQL | Pandas |
|---|---|
| `WHERE Region = 'North'` | `df[df["Region"] == "North"]` |
| `WHERE Total > 3000` | `df[df["Total"] > 3000]` |
| `WHERE A AND B` | `df[(cond_A) & (cond_B)]` |
| `WHERE A OR B` | `df[(cond_A) \| (cond_B)]` |
| `WHERE Region IN ('North','East')` | `df[df["Region"].isin(["North","East"])]` |
| `WHERE Customer LIKE 'A%'` | `df[df["Customer"].str.startswith("A")]` |
| `WHERE X IS NULL` | `df[df["X"].isna()]` |
| `WHERE Total BETWEEN 1000 AND 3000` | `df[df["Total"].between(1000, 3000)]` |

⚠️ In Pandas use `&`, `|`, `~` (not `and`, `or`, `not`) and **wrap each condition in parentheses**.

````python
# ── SQL ──
sql = '''
SELECT OrderID, Customer, Region, Total
FROM orders
WHERE Region = 'North' AND Total > 2000
'''
display(pd.read_sql(sql, con))

# ── Pandas ──
clean.loc[(clean["Region"] == "North") & (clean["Total"] > 2000),
          ["OrderID", "Customer", "Region", "Total"]]
````

````python
# More filters
print("IN      :", len(clean[clean["Region"].isin(["North", "East"])]))
print("LIKE 'A%':", len(clean[clean["Customer"].str.startswith("A")]))
print("BETWEEN :", len(clean[clean["Total"].between(1000, 3000)]))
print("NOT     :", len(clean[~(clean["Product"] == "Laptop")]))
````

## 🔟 SQL `GROUP BY` → Pandas `groupby`

Same idea: *split → aggregate → combine*.

````python
# ── SQL ──
sql = '''
SELECT Region,
       COUNT(*)      AS Orders,
       SUM(Total)    AS Revenue,
       AVG(Total)    AS AvgOrder
FROM orders
GROUP BY Region
ORDER BY Revenue DESC
'''
display(pd.read_sql(sql, con))

# ── Pandas ──
(clean.groupby("Region")
      .agg(Orders=("OrderID", "count"),
           Revenue=("Total", "sum"),
           AvgOrder=("Total", "mean"))
      .sort_values("Revenue", ascending=False)
      .reset_index())
````

````python
# Multiple group columns  (SQL: GROUP BY Region, Product)
clean.groupby(["Region", "Product"])["Total"].sum().reset_index().head(8)
````

````python
# Bonus: Pivot table — the Excel favorite, in one line
pd.pivot_table(clean, index="Region", columns="Product",
               values="Total", aggfunc="sum", fill_value=0, margins=True)
````

````python
# Bonus: a chart in 1 line — something SQL can't do
clean.groupby("Region")["Total"].sum().plot(kind="bar", title="Revenue by region", figsize=(6, 3));
````

## 1️⃣4️⃣ Simple ETL

**E**xtract → **T**ransform → **L**oad

```
 Excel / API / DB   ──►   clean + calculate   ──►   Excel / DB / report
     EXTRACT                 TRANSFORM                    LOAD
```

Good practice: one **function per step**, so the pipeline is readable and reusable.

````python
def extract(path):
    """Read raw data from Excel."""
    return pd.read_excel(path, sheet_name="Orders")

def transform(df):
    """Clean and enrich."""
    df = df.drop_duplicates().copy()                     # remove duplicates
    df["Customer"] = df["Customer"].str.strip().str.title()
    df["Total"] = df["Quantity"] * df["UnitPrice"]       # calculate
    df["Month"] = df["Date"].dt.to_period("M").astype(str)
    return df

def load_excel(df, path):
    df.to_excel(path, index=False)

def load_sqlite(df, con, table):
    df.to_sql(table, con, if_exists="replace", index=False)

# Run the pipeline
raw_data   = extract("sales.xlsx")
final_data = transform(raw_data)
load_excel(final_data, "etl_output.xlsx")
load_sqlite(final_data, con, "orders_clean")

print(f"Extracted {len(raw_data)} rows -> loaded {len(final_data)} clean rows")
pd.read_sql("SELECT COUNT(*) AS rows_in_db FROM orders_clean", con)
````

## 1️⃣5️⃣ Final example: Excel → Python → Transform → Report

**Scenario:** The manager sends `sales.xlsx`. Build a monthly report with:
- cleaned data
- revenue by region and by product
- top customer per region
- a chart
- one Excel file with several sheets, nicely formatted

````python
import matplotlib.pyplot as plt

# 1) EXTRACT
raw = pd.read_excel("sales.xlsx")

# 2) TRANSFORM
data = (raw.drop_duplicates()
           .assign(Total=lambda d: d["Quantity"] * d["UnitPrice"],
                   Month=lambda d: d["Date"].dt.strftime("%Y-%m")))

by_region  = data.groupby("Region", as_index=False)["Total"].sum().sort_values("Total", ascending=False)
by_product = data.groupby("Product", as_index=False)["Total"].sum().sort_values("Total", ascending=False)
by_month   = data.groupby("Month", as_index=False)["Total"].sum()

# Top customer per region
cust = data.groupby(["Region", "Customer"], as_index=False)["Total"].sum()
top_customer = cust.loc[cust.groupby("Region")["Total"].idxmax()].reset_index(drop=True)

# KPIs
kpi = pd.DataFrame({
    "KPI":   ["Orders", "Revenue", "Average order", "Best region"],
    "Value": [len(data), int(data["Total"].sum()), round(data["Total"].mean(), 1), by_region.iloc[0]["Region"]],
})
kpi
````

````python
# 3) CHART
fig, ax = plt.subplots(1, 2, figsize=(11, 3.5))
by_region.plot(kind="bar", x="Region", y="Total", ax=ax[0], legend=False, title="Revenue by region", color="#4C78A8")
by_month.plot(kind="line", x="Month", y="Total", ax=ax[1], marker="o", legend=False, title="Revenue by month", color="#F58518")
plt.tight_layout()
fig.savefig("report_chart.png", dpi=120)
plt.show()
````

````python
# 4) LOAD  -> formatted Excel report
report_file = "Sales_Report.xlsx"
with pd.ExcelWriter(report_file, engine="openpyxl") as xw:
    kpi.to_excel(xw, sheet_name="Summary", index=False)
    by_region.to_excel(xw, sheet_name="By region", index=False)
    by_product.to_excel(xw, sheet_name="By product", index=False)
    by_month.to_excel(xw, sheet_name="By month", index=False)
    top_customer.to_excel(xw, sheet_name="Top customers", index=False)
    data.to_excel(xw, sheet_name="Clean data", index=False)

    # simple formatting: bold header + auto column width
    from openpyxl.styles import Font, PatternFill
    for ws in xw.book.worksheets:
        for cell in ws[1]:
            cell.font = Font(bold=True, color="FFFFFF")
            cell.fill = PatternFill("solid", fgColor="4C78A8")
        for col in ws.columns:
            width = max(len(str(c.value)) if c.value is not None else 0 for c in col) + 3
            ws.column_dimensions[col[0].column_letter].width = width

print("✅ Report created:", report_file)
````

## 🏁 Python vs SQL — when to use what

| Task | Better with |
|---|---|
| Query a huge table (millions+ rows) | **SQL** |
| Join/aggregate inside the database | **SQL** |
| Read Excel / CSV / JSON / API data | **Python** |
| Clean messy data, fix text/dates | **Python** |
| Repeat the same job every day | **Python** |
| Charts, reports, emails, files | **Python** |
| Statistics, machine learning | **Python** |
| Control who sees which data / security | **SQL (database)** |

### Same logic, two languages

| SQL | Pandas |
|---|---|
| `SELECT` | `df[[cols]]` |
| `WHERE` | `df[condition]` |
| `GROUP BY` | `df.groupby()` |
| `ORDER BY` | `sort_values()` |
| `DISTINCT` | `drop_duplicates()` |
| `JOIN` | `pd.merge()` |
| `CASE WHEN` | `np.where()` |
| `COUNT/SUM/AVG` | `count/sum/mean` |

### 🚀 Recommended workflow
**SQL** pulls the right slice of data → **Python** cleans, enriches, visualizes and delivers.
